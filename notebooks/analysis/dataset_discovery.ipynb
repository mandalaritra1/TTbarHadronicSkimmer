```python
from coffea.dataset_tools.rucio_utils import (
    get_rucio_client,
    get_dataset_files_replicas,
)

dataset = "/QCD_Bin-PT-1000to1500_TuneCP5_13p6TeV_pythia8/RunIII2024Summer24NanoAODv15-BTVNanoV15_150X_mcRun3_2024_realistic_v2-v1/NANOAODSIM"

client = get_rucio_client()
```

```python

import json
from pathlib import Path
from coffea.dataset_tools.rucio_utils import get_dataset_files_replicas

output_path = Path("../../data/nanoAOD/QCD_ptbinned.json")

pt_bins = [
    "15to20",
    "20to30",
    "30to50",
    "50to80",
    "80to120",
    "120to170",
    "170to300",
    "300to470",
    "470to600",
    "600to800",
    "800to1000",
    "1000to1500",
    "1500to2000",
    "2000to2500",
    "2500to3000",
]

fileset = {"2024": {}}

for ptbin in pt_bins:
    key = f"QCD_PT-{ptbin}"
    dataset = (
        f"/QCD_Bin-PT-{ptbin}_TuneCP5_13p6TeV_pythia8/"
        f"RunIII2024Summer24NanoAODv15-BTVNanoV15_150X_mcRun3_2024_realistic_v2-v1/"
        f"NANOAODSIM"
    )

    print(f"Fetching {dataset}")

    files, _, _ = get_dataset_files_replicas(
        dataset,
        client = client,
        mode="first",
        partial_allowed=False,
        scope="cms",
    )

    cleaned_files = []
    for f in files:
        if "/store/" in f:
            cleaned_files.append(f[f.index("/store/"):])
        else:
            cleaned_files.append(f)

    fileset["2024"][key] = cleaned_files

output_path.parent.mkdir(parents=True, exist_ok=True)
with open(output_path, "w") as f:
    json.dump(fileset, f, indent=4)

print(f"Saved to {output_path}")
```
